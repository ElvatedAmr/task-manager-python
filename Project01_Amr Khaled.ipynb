#### 2. Simple Quiz Game
>  Create a quiz game that asks the user a series of questions and keeps track of their score.

```python
# Solution - enter your code solution below

quiz = [
    {"question": "Which car manufacturer produces the Mustang?", "answer": "Ford"},
    {"question": "What does 'BMW' stand for?", "answer": "Bayerische Motoren Werke"},
    {"question": "Which Italian car brand has a bull as its logo?", "answer": "Lamborghini"},
    {"question": "In what year was Tesla founded?", "answer": "2003"},
    {"question": "What is the name of Toyota’s luxury car brand?", "answer": "Lexus"}
]

# Starting score counter
score = 0

# Loop the quiz questions
for q in quiz:
    user_answer = input(q["question"] + " ")
#checking if the answers are right
    if user_answer.strip().lower() == q["answer"].lower():
        print("Correct Amazming!")
        score += 1
    else:
        print(f"Oops wrong answer. The correct answer was {q['answer']}.")

print(f"\nYour final score is {score}/{len(quiz)}")
```
